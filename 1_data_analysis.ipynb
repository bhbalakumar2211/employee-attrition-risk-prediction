```python
# 1. Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Load the dataset
print("Loading data...")
df = pd.read_csv('employee_data.csv')

# 3. Basic Data Inspection
print(f"Dataset shape (rows, columns): {df.shape}")
print("\nFirst 3 rows of data:")
display(df.head(3))

# 4. Data Cleaning
# The project guidelines stated some columns are constants and not useful. Let's drop them.
columns_to_drop = ['EmployeeCount', 'EmployeeNumber', 'Over18', 'StandardHours']
df_cleaned = df.drop(columns=columns_to_drop, errors='ignore')

print("\nChecking for missing (null) values:")
print(df_cleaned.isnull().sum())

# Save the cleaned data to a new file so we can use it later
df_cleaned.to_csv('cleaned_employee_data.csv', index=False)
print("\nCleaned data saved as 'cleaned_employee_data.csv'")
```

```python
# 1. Load the cleaned data we just saved
df = pd.read_csv('cleaned_employee_data.csv')

# Set the visual style for our charts
sns.set_style("whitegrid")

# Create a large figure to hold 4 different charts
plt.figure(figsize=(16, 12))

# Chart 1: Overall Attrition Rate
plt.subplot(2, 2, 1)
sns.countplot(x='Attrition', data=df, palette='Set2')
plt.title('Overall Employee Attrition (Yes = Left, No = Stayed)')
plt.ylabel('Number of Employees')

# Chart 2: Attrition by Department
plt.subplot(2, 2, 2)
sns.countplot(x='Department', hue='Attrition', data=df, palette='Set1')
plt.title('Attrition across different Departments')
plt.ylabel('Number of Employees')

# Chart 3: Age Distribution vs Attrition
plt.subplot(2, 2, 3)
sns.histplot(data=df, x='Age', hue='Attrition', kde=True, palette='coolwarm')
plt.title('Age Distribution of Employees (Stayed vs Left)')

# Chart 4: Monthly Income vs Attrition
plt.subplot(2, 2, 4)
sns.boxplot(x='Attrition', y='MonthlyIncome', data=df, palette='pastel')
plt.title('Monthly Income vs Attrition')

# Show all the plots cleanly
plt.tight_layout()
plt.show()

print("EDA Charts generated successfully!")
```
