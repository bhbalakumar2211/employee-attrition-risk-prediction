```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score
import joblib
import warnings
warnings.filterwarnings('ignore')

# 1. Load the cleaned data from Phase 1
df = pd.read_csv('cleaned_employee_data.csv')

# 2. Convert text data into numbers (Machine Learning only understands numbers)
print("Converting text data to numbers...")
label_encoders = {}
for column in df.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    df[column] = le.fit_transform(df[column])
    label_encoders[column] = le

# We MUST save these encoders for our Streamlit app later!
joblib.dump(label_encoders, 'label_encoders.pkl')

# ==========================================
# MODEL 1: Predicting Employee Attrition
# ==========================================
print("\n--- Training Attrition Model ---")
# Separate the 'Attrition' column (what we want to predict) from the rest of the data
X_attrition = df.drop('Attrition', axis=1)
y_attrition = df['Attrition']

# Split data: 80% for training the AI, 20% for testing its accuracy
X_train_a, X_test_a, y_train_a, y_test_a = train_test_split(X_attrition, y_attrition, test_size=0.2, random_state=42)

# Create and train the Random Forest model
rf_attrition = RandomForestClassifier(n_estimators=100, random_state=42)
rf_attrition.fit(X_train_a, y_train_a)

# Test the model and print the project-required metrics (Accuracy, Precision, Recall, F1)
preds_a = rf_attrition.predict(X_test_a)
print(f"Attrition Model Accuracy: {accuracy_score(y_test_a, preds_a) * 100:.2f}%")
print("\nDetailed Report (Precision, Recall, F1-Score):")
print(classification_report(y_test_a, preds_a))

# Save the Attrition model
joblib.dump(rf_attrition, 'attrition_model.pkl')

# ==========================================
# MODEL 2: Predicting Performance Rating
# ==========================================
print("\n--- Training Performance Rating Model ---")
X_perf = df.drop('PerformanceRating', axis=1)
y_perf = df['PerformanceRating']

X_train_p, X_test_p, y_train_p, y_test_p = train_test_split(X_perf, y_perf, test_size=0.2, random_state=42)

rf_perf = RandomForestClassifier(n_estimators=100, random_state=42)
rf_perf.fit(X_train_p, y_train_p)

preds_p = rf_perf.predict(X_test_p)
print(f"Performance Model Accuracy: {accuracy_score(y_test_p, preds_p) * 100:.2f}%")

# Save the Performance model
joblib.dump(rf_perf, 'performance_model.pkl')

print("\n✅ SUCCESS: Models and Encoders saved! You are ready for the Streamlit Dashboard.")
```
